## **Import & Setup**

```python
import json, numpy as np
from transformers import AutoTokenizer
import json, numpy as np

PATH_LLM = "../data/inference_outputs_llm.jsonl"
PATH_RAG = "../data/inference_outputs_rag.jsonl"
MODEL_PATH = "C:\\Users\\Giorgos\\Desktop\\HMMY\\10ο Εξάμηνο\\Διπλωματική\\07. Approach - Fine Tuning\\2. LLM Fine Tuning\\Fine-Tuned LLM Models\\1. Attempt 1\\LLM Model Extrcated\\fine-tuned-mistralai-Mistral-7B-Instruct-v0.2"
```

## **Compare Output Length**

```python
for name, path in [("FT",  PATH_LLM), ("RAG", PATH_RAG)]:
    with open(path, encoding="utf-8") as f:
        outs = [json.loads(l)["model_output"] for l in f]
    chars  = [len(o) for o in outs]
    words  = [len(o.split()) for o in outs]
    empty  = sum(1 for o in outs if not o.strip())
    print(f"{name}: n={len(outs)}  empty={empty}  "
          f"chars mean/median={np.mean(chars):.0f}/{np.median(chars):.0f}  "
          f"words mean/median={np.mean(words):.0f}/{np.median(words):.0f}")
```

## **Compare the Tokenizers**

```python
ft_tok  = AutoTokenizer.from_pretrained(MODEL_PATH)
rag_tok = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3-8B-Instruct")

for name, tok, path in [("FT",  ft_tok,  PATH_LLM), ("RAG", rag_tok, PATH_RAG)]:
    with open(path, encoding="utf-8") as f:
        outs = [json.loads(l)["model_output"] for l in f]
    counts = [len(tok.encode(o, add_special_tokens=False)) for o in outs]
    print(f"{name}: tokens mean/median = {np.mean(counts):.0f}/{np.median(counts):.0f}")
```
